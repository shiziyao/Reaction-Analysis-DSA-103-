```python
def load_and_clean_setup_data(filepath, setup_name):
       """Load and standardize data from one setup"""
       setup_data = pd.read_csv(filepath)
       """convert temperature units to celsius (if necessary)"""
       if setup_name == 'B': 
              setup_data['temp_fahrenheit'] = setup_data['temp_fahrenheit'].apply(lambda x: (x-32) /1.8)
              setup_data["temp_fahrenheit"] = setup_data["temp_fahrenheit"].round(2)
       """Standardize column names"""
       setup_data.columns = ["time (min)", "temperature (°C)", "concentration (M)", "pH", "yield (%)"]
       """handle missing values: if there is a missing value, drop the row"""
       setup_data = setup_data.dropna()
       """add a column to indicate the setup"""
       setup_data['setup']= setup_name


       return setup_data 

print(load_and_clean_setup_data(r"D:\Studium\UZH\BCH\DSA103\Reaction-Analysis-DSA-103-\session06_assignment\data\setup_A.csv", 'A'))
print(load_and_clean_setup_data(r"D:\Studium\UZH\BCH\DSA103\Reaction-Analysis-DSA-103-\session06_assignment\data\setup_B.csv", 'B'))   

setup_A_data = load_and_clean_setup_data(r"D:\Studium\UZH\BCH\DSA103\Reaction-Analysis-DSA-103-\session06_assignment\data\setup_A.csv", 'A')
setup_B_data = load_and_clean_setup_data(r"D:\Studium\UZH\BCH\DSA103\Reaction-Analysis-DSA-103-\session06_assignment\data\setup_B.csv", 'B')
```

```python
def combine_datasets(setup_a_data, setup_b_data):
       """Combine datasets from both setups"""
       combined_data= pd.concat([setup_a_data, setup_b_data], ignore_index=True)
       return combined_data

print(combine_datasets(setup_A_data, setup_B_data))
```
